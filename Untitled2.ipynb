```python
import pandas as pd

# Replace with your actual file name if it's different
df = pd.read_csv('fake reviews dataset.csv')

# Show the first few rows
df.head()
```

```python
# Display the column names in your dataset
print(df.columns)
```

```python
# Extract review text and labels
X = df['text_']  # Use the 'text_' column for reviews
y = df['label']  # Use the 'label' column for labels (real or fake)

# Convert text to numerical features using TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)
X_vec = vectorizer.fit_transform(X)

# Show the shape of the transformed data
print(X_vec.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_vec, y, test_size=0.2, random_state=42)

# Create and train the Logistic Regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
# Example: Test the model with a custom review
sample_review = ["verry much recomended "]
sample_vec = vectorizer.transform(sample_review)  # Convert text to features

# Predict the label (fake or real)
pred = model.predict(sample_vec)

# Output the result
print("Prediction:", "Fake" if pred[0] == 1 else "Real")
```

```python
import joblib

# Save the trained model
joblib.dump(model, 'fake_review_model.pkl')

# Save the vectorizer too
joblib.dump(vectorizer, 'vectorizer.pkl')

print("Model and vectorizer saved successfully.")
```

```python
# Load the model and vectorizer
model = joblib.load('fake_review_model.pkl')
vectorizer = joblib.load('vectorizer.pkl')

# Now you can predict new reviews using the loaded model
sample_review = ["This is the best product ever, it will change your life!"]
sample_vec = vectorizer.transform(sample_review)
pred = model.predict(sample_vec)

print("Prediction:", "Fake" if pred[0] == 1 else "Real")
```

```python
# Adjust class weights to handle imbalance
from sklearn.linear_model import LogisticRegression

# Set the class_weight='balanced' option to handle imbalanced dataset
model = LogisticRegression(class_weight='balanced')

# Train the model again
model.fit(X_train, y_train)

# Predict and evaluate the model
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Reduce dataset size (10% of original training data)
X_train_small, _, y_train_small, _ = train_test_split(X_train, y_train, train_size=0.1, random_state=42)

# Create and train a Logistic Regression model with class_weight='balanced' to handle imbalance
lr_model = LogisticRegression(class_weight='balanced', max_iter=500, random_state=42)

# Train the model on the smaller dataset
lr_model.fit(X_train_small, y_train_small)

# Predict on the test set
y_pred_lr = lr_model.predict(X_test)

# Print accuracy and classification report
print("Accuracy (Logistic Regression):", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report (Logistic Regression):\n", classification_report(y_test, y_pred_lr))
```
